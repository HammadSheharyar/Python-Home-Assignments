```python
"""
Loop Through Dictionary and Remove Items
Given:
students = {"Ali": 45, "Sara": 90, "Rohan": 29, "Meera": 50}

Remove all students scoring less than 40
 without modifying the dictionary while looping over it.

"""
students = {"Ali": 45, "Sara": 90, "Rohan": 29, "Meera": 50}
for name, score in students.items():
  if score < 40:
    del students[name]

print(students)
```
