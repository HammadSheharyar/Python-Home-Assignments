```python
import sympy as sp
```

```python
a_, b_ , c_ = sp.symbols("a b c")

eq1 = sp.Eq(a_ + b_ + c_, 3)
eq2 = sp.Eq(4*a_ + 2*b_ + c_, 4)
eq3 = sp.Eq(9*a_ + 3*b_ + c_, 11)
print(eq1)

sol = sp.solve((eq1, eq2, eq3), (a_, b_, c_))
print(sol)
a, b , c = sol[a_],sol[b_] , sol[c_]
print(a, b , c)
```
