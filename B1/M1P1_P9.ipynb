```python
"""
Ask the user to enter the first four terms of a sequence.
Store them in a list.

"""

seq = []
for i in range(1, 5):
  term = int(input(f"Enter the {i} term: "))
  seq.append(term)
```

```python
print(seq)
```

```python
first_diff = []
for i in range(3):
  d = seq[i + 1] - seq[i]
  first_diff.append(d)
print(first_diff)
```

```python
first_diff = [seq[i + 1] - seq[i] for i in range(3)]
print(first_diff)
```

```python
if all([d == first_diff[0] for d in first_diff]):
  print("Linear Sequence")

else:
  print("not linear")
```

```python
sec_diff = [first_diff[i+1] - first_diff[i] for i in range(2)]
print(sec_diff)
```

```python
import sympy as sp
a_, b_, c_ = sp.symbols("a b c")
eq1 = sp.Eq(a_ + b_ + c_ , 3)
eq2 = sp.Eq(4*a_ + 2*b_ + c_ , 4)
eq3 = sp.Eq(9*a_ + 3*b_ + c_ , 11)
sol = sp.solve((eq1, eq2, eq3), (a_, b_, c_))
print(sol)
a, b , c = sol[a_], sol[b_], sol[c_]
print(a, b ,c)
```

```python
list1 = [0, 2, 8, -1, -3, 1]
# make a ftn in which find the largest number from the given list

def find_largest(numbers):
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest
list1 = [0, 2, 8, -1, -3, 1]
print("Largest number is:", find_largest(list1))
```
