```python
# x**2 + 1
# lamdify function
x = symbols("x")
expr = x**2 + 1
f = sp.lambdify(x, expr, "numpy")
x_val = np.linspace(-10, 10, 0)
y_val = f(x_val)
# print(x_val)
# print(y_val)

plt.plot(x_val, y_val)
```

```python
expr2 = x**2 - 3*x + 2
f = sp.lambdify(x , expr2, "numpy")
y_val = f(x_val)
plt.plot(x_val, y_val, label = "x**2 - 3*x + 2")
plt.legend()
plt.show()
```
