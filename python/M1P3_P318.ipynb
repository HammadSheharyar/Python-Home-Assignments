```python
"""
Create a Python program that acts as a Math Puzzle Builder.
The program should:
Present 3–4 different math puzzles (addition, subtraction, multiplication, division, or comparing numbers).


Ask the user to answer each puzzle.


Keep track of correct answers and calculate a score at the end.


Display a final message based on the score.




"""
# Your output should be like this
"""
🧩 Example Output
Puzzle 1: Addition
Enter first number: 5
Enter second number: 3
What is 5.0 + 3.0? 8
✅ Correct!

Puzzle 2: Subtraction
Enter first number: 9
Enter second number: 4
What is 9.0 - 4.0? 6
❌ Wrong!

Puzzle 3: Multiplication
Enter first number: 2
Enter second number: 6
What is 2.0 × 6.0? 12
✅ Correct!

Puzzle 4: Compare Numbers
Enter first number: 7
Enter second number: 7
Which number is greater? (a/b/equal): equal
✅ Correct!

🏆 Your Score: 3 / 4
👍 Good Job! Keep Practicing!

"""
```

```python
import random
print("Welcome to the Simplest Math Puzzle Builder!")
score = 0
num_puzzles_total = 3
# --- Puzzle 1: Addition ---
print("\n--- Puzzle 1: Addition ---")
num1_add = round(random.uniform(1, 10), 0)
num2_add = round(random.uniform(1, 10), 0)
correct_answer_add = num1_add + num2_add
question_add = f"What is {num1_add} + {num2_add}? "
user_answer_add = float(input(question_add))
if user_answer_add == correct_answer_add:
    print("Correct!")
    score += 1
else:
    print("Wrong!")
    print(f"The correct answer was {correct_answer_add}")
# --- Puzzle 2: Subtraction ---
print("\n--- Puzzle 2: Subtraction ---")
num1_sub = round(random.uniform(1, 100), 0)
num2_sub = round(random.uniform(1, 100), 0)
correct_answer = num1_sub - num2_sub
question_sub = f"What is {num1_sub} - {num2_sub}? "
user_answer = float(input(question_sub))
if user_answer== correct_answer:
    print("Correct!")
    score += 1
else:
    print("Wrong!")
    print(f"The correct answer was {correct_answer}")
# --- Puzzle 3: Multiplication ---
print("\n--- Puzzle 3: Multiplication ---")
num1 = round(random.uniform(1, 100), 0)
num2 = round(random.uniform(1, 100), 0)
correct_answer = num1 * num2
question_mul = f"What is {num1} * {num2}? "
user_answer= float(input(question_mul))
if user_answer == correct_answer:
    print("Correct!")
    score += 1
else:
    print("Wrong!")
    print(f"The correct answer was {correct_answer}")

print(f"\nYour Final Score: {score} / {num_puzzles_total}")
if score == num_puzzles_total:
    print("Perfect Score! Amazing!")
elif score > 0:
    print("Good Effort! Keep practicing!")
else:
    print("You'll get it next time!")
```
