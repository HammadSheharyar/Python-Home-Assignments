```python
def sec_greatest(number_list):
  """
  Finds the largest number in a list of numbers without using any
  built-in functions
  """
  greatest_num = number_list[0]
  for current_number in number_list[1:]:
    if current_number >  greatest_num:
       greatest_num = current_number

  return greatest_num
```
